```python
import pandas as pd
```

```python
! .\.venv10\Scripts\activate
from pycaret.classification import evaluate_model, tune_model, setup, compare_models, blend_models, stack_models
```

```python
data = pd.read_csv('../data/processed_smoke_detector.csv')
data.head()
```

```python
exp = setup(data=data,
            target='Fire Alarm', 
            normalize=True, 
            train_size=0.8, 
            session_id=42)
```

```python
best_model = compare_models()
```

```python
tuned_model = tune_model(best_model)
```

```python
evaluate_model(tuned_model)
```

```python
top3_models = compare_models(n_select=3)
blended_model = blend_models(estimator_list=top3_models)
evaluate_model(blended_model)
```

```python
top5_models = compare_models(n_select=5)
stacked_model = stack_models(estimator_list=top5_models, meta_model=None)
evaluate_model(stacked_model)
```
