```python
import numpy as np
import copy
```

```python
def Kmeans(array, k, max_iter=1000, tol=1e-4):
    data = np.array(array)
    n = data.shape[0]
    
    indices = np.random.choice(n, size=k, replace=False)
    centroids = data[indices].tolist()
    
    for _ in range(max_iter):
        labels = data_distribution(data, centroids)
        
        old_centroids = copy.deepcopy(centroids)
        
        centroids = cluster_update(data, labels, k)
        
        if np.allclose(old_centroids, centroids, atol=tol):
            break
    
    return labels

def data_distribution(data, centroids):
    return [np.argmin([np.linalg.norm(point - c) for c in centroids]) 
            for point in data]

def cluster_update(data, labels, k):
    new_centroids = []
    for i in range(k):
        cluster_points = data[labels == i]
        if len(cluster_points) == 0:
            new_centroids.append(data[np.random.randint(len(data))].tolist())
            continue
        new_centroids.append(cluster_points.mean(axis=0).tolist())
    return new_centroids
```

```python
import pandas as pd
```

```python
from sklearn.metrics import calinski_harabasz_score
from sklearn.metrics import adjusted_rand_score
from sklearn.metrics.cluster import rand_score
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler
```

```python
data = pd.read_csv('../data/processed_smoke_detector.csv')
print(data.head())
data_cl = data.drop('Fire Alarm', axis=1)
data_y_cl = data['Fire Alarm']
scal_data_cl = StandardScaler().fit_transform(data_cl)
```

```python
labels = Kmeans(scal_data_cl, 4, 100)
```

```python
print("Внешние метрики:")
print("rand score", rand_score(labels, data_y_cl))
print("adjusted_rand_score", adjusted_rand_score(data_y_cl, labels))
print("Внутренние метрики:")
print("silhouette_score: ", silhouette_score(scal_data_cl, labels))
print("calinski_harabasz_score", calinski_harabasz_score(scal_data_cl, labels))
```
